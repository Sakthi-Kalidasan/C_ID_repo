## Understanding Undershoot Requirements

```python
V_min = 1.1  # Minimum Voltage in Volts
V_target = 1.2  # Target Voltage in Volts
C_Load = 50e-12  # Load Capacitance in Farads
I_Load = 10e-3  # Load Current in Amperes

V_undershoot = V_target - V_min  # Voltage Undershoot in Volts

t_response = (C_Load * V_undershoot) / I_Load  # Response Time in seconds
print(f"Response Time: {t_response * 1e9:.2f} ns")
```
